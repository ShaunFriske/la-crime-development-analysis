# Begin Data Cleaning: Duplicates, Formats, and Missing Values

```python
# Create a new copy to store our cleaned data, keeping the raw data safe
cleaned_crime_df = raw_crime_df.copy()
```

```python
# Drop the duplicate rows you found earlier
cleaned_crime_df = cleaned_crime_df.drop_duplicates()
```

```python
# Fix the date format and explicitly tell Pandas to handle mixed date strings
if 'DATE OCC' in cleaned_crime_df.columns:
    cleaned_crime_df['DATE OCC'] = pd.to_datetime(cleaned_crime_df['DATE OCC'], format='mixed', errors='coerce')
```

```python
# Drop rows missing crucial geographic data (location is mandatory for real estate analysis)
cleaned_crime_df = cleaned_crime_df.dropna(subset=['AREA NAME', 'LAT', 'LON'])
```

```python
# Keep only rows with logical ages (between 0 and 120)
if 'Vict Age' in cleaned_crime_df.columns:
    cleaned_crime_df = cleaned_crime_df[(cleaned_crime_df['Vict Age'] >= 0) & (cleaned_crime_df['Vict Age'] <= 120)]
```

```python
# Remove the erroneous 0.0 coordinates you found in the inspection phase
if 'LAT' in cleaned_crime_df.columns and 'LON' in cleaned_crime_df.columns:
    cleaned_crime_df = cleaned_crime_df[(cleaned_crime_df['LAT'] != 0.0) & (cleaned_crime_df['LON'] != 0.0)]
```

```python
# Define a list of columns that do not help determine general neighborhood safety
cols_to_drop = ['DR_NO', 'Cross Street', 'Crm Cd 1', 'Crm Cd 2', 'Crm Cd 3', 'Crm Cd 4']
# Check which of these columns actually exist to avoid code errors, then drop them
existing_cols = [col for col in cols_to_drop if col in cleaned_crime_df.columns]
cleaned_crime_df = cleaned_crime_df.drop(columns=existing_cols)
```

# Analysis Part 1: Categorizing Crimes (Property vs. Violent)

**Why this step:** LAPD data contains hundreds of highly specific crime codes. To estimate security costs and tenant safety, we must group these into two broad, manageable categories: Property Crime and Violent Crime.
**What we are doing:** We are creating a new column called `Crime Category`. We will search the crime description column for keywords (e.g., 'BURGLARY', 'ASSAULT') and assign the appropriate category.

```python
# Ensure the description column is treated as text and make it uppercase for easy searching
cleaned_crime_df['Crm Cd Desc'] = cleaned_crime_df['Crm Cd Desc'].astype(str).str.upper()

# Define keywords for our business case
property_keywords = 'BURGLARY|VANDALISM|THEFT|STOLEN|TRESPASS'
violent_keywords = 'ASSAULT|ROBBERY|BATTERY|HOMICIDE|KIDNAPPING|RAPE'

# Create conditions to check if the description contains our keywords
conditions = [
    cleaned_crime_df['Crm Cd Desc'].str.contains(violent_keywords, na=False),
    cleaned_crime_df['Crm Cd Desc'].str.contains(property_keywords, na=False)
]

# Assign the categories based on the conditions above
choices = ['Violent Crime', 'Property Crime']

# Create the new column. If a crime doesn't fit either, it gets labeled 'Other'
cleaned_crime_df['Crime Category'] = np.select(conditions, choices, default='Other')
```

```python
# Display the counts of our new categories to ensure the categorization worked
cleaned_crime_df['Crime Category'].value_counts()
```

Analysis Part 2: Neighborhood Safety (Property & Violent Rates)

**Why  this step:** We need to judge project marketability, tenant safety, and insurance costs on a neighborhood-by-neighborhood basis.
**What we are doing:** We are grouping the data by `AREA NAME` (LAPD's term for neighborhoods) and counting the number of Property and Violent crimes in each area to identify the lowest-risk areas.

```python
# Create a pivot table to count Property vs Violent crimes per neighborhood
neighborhood_safety_comparison = pd.crosstab(
    index=cleaned_crime_df['AREA NAME'],
    columns=cleaned_crime_df['Crime Category']
)

# Sort the table so the neighborhoods with the lowest Violent Crime are at the top
neighborhood_safety_comparison = neighborhood_safety_comparison.sort_values(by='Violent Crime', ascending=True)

# Display the safety comparison table
neighborhood_safety_comparison
```

# Analysis Part 4: Identifying Micro-Hotspots (Street Segments)

**Why this step:** For design and planning, we need to know exactly which blocks or street segments pose the highest risk to dictate where we place secure parking, lighting, and entry points.
**What we are doing:** We will count the frequencies of crimes at the exact string locations provided in the `LOCATION` column to find the worst micro-hotspots in the city.

```python
# Group the dataset by the exact street location and count the occurrences
street_segment_hotspots = cleaned_crime_df['LOCATION'].value_counts().reset_index()

# Rename the columns so they are easily readable for a real estate report
street_segment_hotspots.columns = ['Street Segment Location', 'Total Historical Crimes']

# Display the top 20 most dangerous individual street segments in the dataset
street_segment_hotspots.head(20)
```
